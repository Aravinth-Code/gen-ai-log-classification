```python
from sentence_transformers import SentenceTransformer
from sklearn.cluster import DBSCAN

# load pretrained
model = SentenceTransformer("all-MiniLM-L6-v2")

# generate embeddings
embeddings = model.encode(dataframe["log_message"])
embeddings.shape
```

```python
# perform dbscan clustering
dbscan = DBSCAN(eps=0.2, min_samples=5, metric="euclidean")
clusters = dbscan.fit_predict(embeddings)

dataframe["cluster"] = clusters
dataframe.head()
```

```python
dataframe["cluster"].unique()
```

```python
dataframe[dataframe.cluster == 1].head()
```

```python
cluster_counts = dataframe["cluster"].value_counts()
large_clusters = cluster_counts[cluster_counts > 10].index

for cluster in large_clusters:
    print(f"Cluster {cluster}:")
    print(dataframe[dataframe["cluster"] == "cluster"]["log_message"].head(5).to_string(index=False))
    print()
```

```python
import re
def classify_with_regex(log_message):
    regex_patterns = {
        r"User User\d+ logged (in|out).": "User Action",
        r"Backup (started|ended) at .*": "System Notification",
        r"Backup completed successfully.": "System Notification",
        r"System updated to version .*": "System Notification",
        r"File .* uploaded successfully by user .*": "System Notification",
        r"Disk cleanup completed successfully.": "System Notification",
        r"System reboot initiated by user .*": "System Notification",
        r"Account with ID .* created by .*": "User Action"
    }
    for pattern, label in regex_patterns.items():
        if re.search(pattern, log_message):
            return label
    return None
```

```python
classify_with_regex("User User123 logged in.")
```

```python
classify_with_regex("System reboot initiated by user User179.")
```

```python
classify_with_regex("Hey you, chill bro")
```

```python
# Apply regex classification
dataframe['regex_label'] = dataframe['log_message'].apply(lambda x: classify_with_regex(x))
dataframe[dataframe['regex_label'].notnull()]
```

```python
dataframe[dataframe['regex_label'].isnull()].head(5)
```

```python
dataframe_non_regex = dataframe[dataframe['regex_label'].isnull()].copy()
dataframe_non_regex.shape
```

```python
print("Before regex filtering:", dataframe.shape)
print("After regex filtering:", dataframe_non_regex.shape)
```

```python
print(dataframe_non_regex["target_label"].value_counts()[dataframe_non_regex["target_label"].value_counts() <= 5].index.tolist())
```

```python
dataframe_non_legacy = dataframe_non_regex[dataframe_non_regex.source != "LegacyCRM"]
dataframe_non_legacy.source.unique()
```

```python
filtered_embeddings = model.encode(dataframe_non_legacy["log_message"].tolist())
filtered_embeddings[:2]
```

```python
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

# Split the data into train and test sets
X_train, X_test, y_train, y_test = train_test_split(
    filtered_embeddings, 
    dataframe_non_legacy["target_label"], 
    test_size=0.2, 
    random_state=42
)

# Train logistic regression model
log_reg = LogisticRegression(max_iter=1000, random_state=42)
log_reg.fit(X_train, y_train)

# Evaluate on test set
train_score = log_reg.score(X_train, y_train)
test_score = log_reg.score(X_test, y_test)
y_predict = log_reg.predict(X_test)

report = classification_report(y_test, y_predict)
print(report)
```
