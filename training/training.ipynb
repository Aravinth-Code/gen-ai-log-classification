```python
import pandas as pd
dataframe = pd.read_csv(r"E:\Projects\gen-ai-log-classification\training\dataset\synthetic_logs.csv")
dataframe.head()
```

```python
dataframe.source.unique()
```

```python
dataframe.target_label.unique()
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.cluster import DBSCAN

# load pretrained
model = SentenceTransformer("all-MiniLM-L6-v2")

# generate embeddings
embeddings = model.encode(dataframe["log_message"])
embeddings.shape
```

```python
# perform dbscan clustering
dbscan = DBSCAN(eps=0.2, min_samples=5, metric="euclidean")
clusters = dbscan.fit_predict(embeddings)

dataframe["cluster"] = clusters
dataframe.head()
```
